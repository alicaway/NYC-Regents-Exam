```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import AutoClean as ac
```

2. Load the Dataset

```python
df = pd.read_csv('2014-15_to_2017-19_NYC_Regents_Exam_Results_-_Public_20260724.csv')
df
```

```python
df.info()
df.dtypes
df.isna().sum()
```

```python
# How many rows are suppressed ('s') in our target column?
print(df['Percent Scoring 65 or Above'].value_counts().head())
print("\nSuppressed rows:", (df['Percent Scoring 65 or Above'] == 's').sum())
```

```python
# Run AutoClean to clean the data
cleaning_pipeline = ac.AutoClean(df)
df_clean = cleaning_pipeline.output
df_clean
```

```python
df_clean.shape
df_clean.columns
df_clean.head(10)
df_clean['Percent Scoring 65 or Above']
```

```python
# Eliminate rows with suppressed values in the target column
df_clean = df_clean[df_clean['Percent Scoring 65 or Above'] != 's']
df_clean["Percent Scoring 65 or Above"] = df_clean["Percent Scoring 65 or Above"].astype("float")

# Create a target column called "Pass_Target" that is 1 if the student passed (scored 65 or above) and 0 if they failed (scored below 65)
df_clean['Pass_Target'] = (df_clean['Percent Scoring 65 or Above'] >= 65).astype(int)
df_clean
```
